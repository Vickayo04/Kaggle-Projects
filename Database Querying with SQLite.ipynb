```python
import csv, sqlite3

con = sqlite3.connect("FinalDB.db")
cur = con.cursor()
```

```python
%load_ext sql
```

```python
import pandas

df1 = pandas.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCensusData.csv?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDB0201ENSkillsNetwork20127838-2021-01-01")
df1.to_sql("Chicago_census", con, if_exists='replace', index=False, method="multi")

df2 = pandas.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoPublicSchools.csv?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDB0201ENSkillsNetwork20127838-2021-01-01")
df2.to_sql("Chicago_schools", con, if_exists='replace', index=False, method="multi")

df3 = pandas.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCrimeData.csv?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDB0201ENSkillsNetwork20127838-2021-01-01")
df3.to_sql("Chicago_crime", con, if_exists='replace', index=False, method="multi")
```

```python
%sql sqlite:///FinalDB.db
```

 **Querying the Database using SQLite**

```python
%sql select count(*) from Chicago_crime
```

```python
%sql select * from Chicago_crime limit 15
```

```python
%sql SELECT COMMUNITY_AREA_NUMBER, COMMUNITY_AREA_NAME FROM Chicago_census WHERE PER_CAPITA_INCOME<11000
```

```python
%sql SELECT CASE_NUMBER FROM Chicago_crime WHERE "PRIMARY_TYPE" = "OFFENSE INVOLVING CHILDREN"
```

```python
%sql SELECT * FROM Chicago_crime WHERE PRIMARY_TYPE = "KIDNAPPING" AND DESCRIPTION LIKE "%CHILD%"
```

```python
%sql select distinct(PRIMARY_TYPE) from Chicago_crime
```

```python
%sql select "Elementary, Middle, or High School", avg(SAFETY_SCORE) from Chicago_schools group by "Elementary, Middle, or High School"
```

```python
%sql select COMMUNITY_AREA_NAME FROM Chicago_census ORDER BY PERCENT_HOUSEHOLDS_BELOW_POVERTY DESC LIMIT 5
```

```python
%sql select COMMUNITY_AREA_NAME, COMMUNITY_AREA_NUMBER from Chicago_census WHERE HARDSHIP_INDEX = (SELECT MAX(HARDSHIP_INDEX) FROM Chicago_census)
```

```python
%sql select COUNT(COMMUNITY_AREA_NUMBER) from Chicago_census group by COMMUNITY_AREA_NUMBER
```

```python
%%sql
SELECT COMMUNITY_AREA_NAME, COMMUNITY_AREA_NUMBER
FROM Chicago_census
WHERE COMMUNITY_AREA_NUMBER = (
    SELECT COMMUNITY_AREA_NUMBER
    FROM Chicago_crime
    GROUP BY COMMUNITY_AREA_NUMBER
    ORDER BY COUNT(ID) DESC
    LIMIT 1
);
```
